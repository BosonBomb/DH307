```python
import cv2
import numpy as np
import os
l =[]
def process_images(filename, image_file, annotation_file, output_dir):
    # Define colors for the two classes
    colors = {
        '0': (0,1,0),  # Green for Class 1
        '1': (0,2,0)  # Blue for Class 2
    }
    
    # Read the image
    image = cv2.imread(image_file)
    # print(image)
    height, width = image.shape[:2]
    
    # Initialize a blank mask image
    
    
    # Read and parse the annotation file

    with open(annotation_file, 'r') as file:
        
        for line in file:
            
            elements = line.strip().split()
            
            # print(elements)
            # elements = elements[0]
            class_label = elements[0]  # Class label is the first element
            coordinates = elements[1:]
            
            if class_label not in colors:
                continue  # Skip classes that are not defined in the colors dictionary

            # Convert coordinates to float and then to pixel values
            coordinates = np.array(coordinates).reshape(-1, 2).astype(float)
            coordinates = np.clip(coordinates, 0, 1)  # Ensure coordinates are within [0, 1]
            polygons = np.array([[int(x * width), int(y * height)] for x, y in coordinates], dtype=np.int32)
            # print(polygons)
            # Draw the polygon on the mask image
            color = colors[class_label]
            mask = np.zeros_like(image)
            mask= cv2.fillPoly(mask, [polygons], color)
            mask = cv2.cvtColor(mask, cv2.COLOR_BGR2GRAY)
    
    # Save the mask image
    labelled_image = os.path.join(output_dir, filename)
    l.append(mask)
    cv2.imwrite(labelled_image,mask)

def generate_masks_for_dataset(image_dir, annotation_dir, output_dir):
    """
    Iterate through the dataset and create masks for each image.
    """
    # Ensure output directory exists
    if not os.path.exists(output_dir):
        os.makedirs(output_dir)
    
    for filename in os.listdir(image_dir):
        if filename.endswith('.jpg'):
            image_file = os.path.join(image_dir, filename)
            annotation_file = os.path.join(annotation_dir, filename.replace('.jpg', '.txt'))

            if os.path.exists(annotation_file):
                # print(1)
                process_images(filename, image_file, annotation_file, output_dir)
            else:
                print(f"Annotation file missing for {filename}")

# Define your paths
image_dir = 'test_practice/test_practice'
annotation_dir = 'test_practice/test_practice_labels'
output_dir = 'labelled_images'

generate_masks_for_dataset(image_dir, annotation_dir, output_dir)
```
