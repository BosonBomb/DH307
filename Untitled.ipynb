```python
import numpy as np

def polygon_to_bbox(polygon):
    """
    Convert polygon segmentation to bounding box.
    
    Args:
        polygon (list of tuples): List of (x, y) coordinates defining the polygon.
    
    Returns:
        bbox (tuple): Bounding box (x_min, y_min, x_max, y_max).
    """
    # Extract all x and y coordinates from the polygon
    x_coords = [point[0] for point in polygon]
    y_coords = [point[1] for point in polygon]
    
    # Get the min and max values for x and y coordinates
    x_min = min(x_coords)
    y_min = min(y_coords)
    x_max = max(x_coords)
    y_max = max(y_coords)
    
    return (x_min, y_min, x_max, y_max)
```

```python
import os
import pandas as pd

def process_folder(input_folder, output_file):
    bbox_data = []

    # Iterate over all .txt files in the folder
    for filename in os.listdir(input_folder):
        if filename.endswith(".txt"):
            file_path = os.path.join(input_folder, filename)
            with open(file_path, 'r') as file:
                for line in file:
                    parts = line.strip().split()
                    
                    # Extract the image identifier (if applicable)
                    img_id = filename.replace(".txt", "")
                    
                    # Extract normalized coordinates
                    coords = list(map(float, parts[1:]))
                    
                    # Convert coordinates to list of (x, y) tuples
                    polygons = [(coords[i], coords[i + 1]) for i in range(0, len(coords), 2)]
                    
                    # Convert polygons to bounding boxes
                    bboxes = [polygon_to_bbox(polygons)]
                    
                    # Flatten the list of bounding boxes to store in CSV
                    bbox_str = ";".join([f"{x_min},{y_min},{x_max},{y_max}" for (x_min, y_min, x_max, y_max) in bboxes])
                    
                    bbox_data.append([img_id, bbox_str])

    # Create a DataFrame and save to CSV
    bbox_df = pd.DataFrame(bbox_data, columns=['image_id', 'bounding_boxes'])
    bbox_df.to_csv(output_file, index=False)
    print(f"Bounding boxes saved to {output_file}")

# Example usage
input_folder = "test_practice/test_practice_labels"
output_file = "bboxes.csv"
process_folder(input_folder, output_file)
```
